```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Disney", "Princess", "prince", "colgate"],
    "Gender": ["Male", "Female", "Male", "Female"],
    "Age": [21, 21, 19, 20]
}

df = pd.DataFrame(data)

df
```

```python
df = pd.get_dummies(df, columns=["Gender"], dtype=int)

df
```

```python
df = df.drop("Name", axis=1)

df
```

```python
arr = df.to_numpy()

arr
```

```python
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Dimension:", arr.ndim)
```

```python
A = arr[:2, :]
B = arr[2:, :]

print("A:")
print(A)

print("B:")
print(B)
```

```python
A + B
```

```python
A * B
```

```python
A - B
```

```python
np.dot(A, B.T)
```
